```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
```

```python
# tool create

@tool
def multiply(a: int, b: int) -> int:
  """Given 2 numbers a and b this tool returns their product"""
  return a * b
```

```python
from langchain_core.tools import tool
import requests

@tool
def get_conversion_factor(base_currency: str, target_currency: str) -> float:
    """Fetch the currency conversion factor between a base currency and a target currency."""
    url = f"https://v6.exchangerate-api.com/v6/c754eab14ffab33112e380ca/pair/{base_currency}/{target_currency}"
    response = requests.get(url)
    data = response.json()
    # Return the numeric rate directly so the LLM gets a clear number
    return data.get("conversion_rate")

@tool
def convert(base_currency_value: float, conversion_rate: float) -> float:
    """Calculate the target currency value from a base amount and a conversion rate."""
    return base_currency_value * conversion_rate
```

```python
convert.args
```

```python
get_conversion_factor.invoke({'base_currency':'USD','target_currency':'INR'})
```

```python
convert.invoke({'base_currency_value':10, 'conversion_rate':85.16})
```

```python
llm_with_tools = llm.bind_tools([get_conversion_factor, convert])
```

```python
messages = [HumanMessage('What is the conversion factor between INR and USD, and based on that can you convert 10 inr to usd')]
```

```python
from langchain.agents import create_agent

# 1. Define your tools
tools = [get_conversion_factor, convert]

# 2. Create the agent runtime directly
agent = create_agent(
    model=llm,
    tools=tools
)

# 3. Run the agent
response = agent.invoke({
    "messages": [("user", "Convert 100 USD to EUR")]
})

# 4. Access the final answer
print(response["messages"][-1].content)
```
